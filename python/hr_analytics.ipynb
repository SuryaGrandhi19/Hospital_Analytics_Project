```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns.tolist()
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.strip()
```

```python
df["date_of_admission"] = pd.to_datetime(
    df["date_of_admission"],
    errors="coerce"
)
```

```python
df["discharge_date"] = pd.to_datetime(
    df["discharge_date"],
    errors="coerce"
)
```

```python
df["length_of_stay"] = (
    df["discharge_date"] - df["date_of_admission"]
).dt.days
```

```python
df[[
    "date_of_admission",
    "discharge_date",
    "length_of_stay"
]].head()
```

```python
df[[
    "date_of_admission",
    "discharge_date",
    "length_of_stay"
]].head()
```

```python
bins = [0, 18, 30, 45, 60, 100]

labels = [
    "0-18",
    "19-30",
    "31-45",
    "46-60",
    "60+"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
df[["age", "age_group"]].head(10)
```

```python
df["admission_year"] = df["date_of_admission"].dt.year
```

```python
df["admission_month"] = df["date_of_admission"].dt.month
```

```python
df["admission_month_name"] = (
    df["date_of_admission"].dt.strftime("%B")
)
```

```python
revenue_by_condition = (
    df.groupby("medical_condition")["billing_amount"]
      .sum()
      .sort_values(ascending=False)
)

revenue_by_condition
```

```python
monthly_admissions = (
    df.groupby(
        ["admission_month", "admission_month_name"]
    )
    .size()
    .reset_index(name="patients")
    .sort_values("admission_month")
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_admissions["admission_month_name"],
    monthly_admissions["patients"],
    marker="o"
)

plt.title("Monthly Hospital Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "../images/monthly_hospital_admissions.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
monthly_revenue = (
    df.groupby(
        ["admission_month", "admission_month_name"]
    )["billing_amount"]
    .sum()
    .reset_index()
    .sort_values("admission_month")
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue["admission_month_name"],
    monthly_revenue["billing_amount"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "../images/monthly_revenue_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
